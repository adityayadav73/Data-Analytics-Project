```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("ecommerce_sales_data.csv")
print(df.info())
print(df.head(10))

```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])

df['Year'] = df['Order Date'].dt.year
yearly_sales = df.groupby(['Year','Region']) ['Sales'].sum().reset_index()

print(yearly_sales)
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'],errors='coerce')


df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month_name()

df_2022 =df[df['Year'] == 2022]

monthly_sales_2022 = df_2022.groupby('Month')['Sales'].sum().reset_index()

print(monthly_sales_2022)
```

```python
month = monthly_sales_2022['Month']
sales = monthly_sales_2022['Sales']
plt.bar(month,sales,color ='coral', width=0.4,label='2024-sales-product')
plt.xticks(rotation=45)
plt.title('Ecommerce_sales_date-2022')
plt.ylabel('Sales')
plt.xlabel('Month')
plt.tight_layout()
#plt.legend()
plt.savefig("Product_Name_sales_2024.png")
plt.show()
```
