```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("employee_income.csv")
print(df.info())
#print(df.head(10))
print(df.tail(10))
```

```python
experience =df['experience']
income = df['income']
plt.scatter(experience,income,color='blue',marker='s',label='experience_income')
plt.title('Employee_income')
plt.xlabel('Experience')
plt.ylabel('Income')
plt.legend()
plt.show()
```
